```python
import math
import json
import os
import datetime

# Function to calculate the area of a rectangle
def calculate_rectangle_area(length, width):
    if length <= 0 or width <= 0:
        raise ValueError("Length and width must be greater than zero")
    return length * width

# Function to calculate the perimeter of a rectangle
def calculate_rectangle_perimeter(length, width):
    if length <= 0 or width <= 0:
        raise ValueError("Length and width must be greater than zero")
    return 2 * (length + width)

# Function to check if a number is even or odd
def is_even(number):
    return number % 2 == 0

# Function to convert a string to uppercase
def to_uppercase(s):
    return s.upper()

# Function to read a JSON file and return its content as a dictionary
def read_json(file_path):
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"{file_path} does not exist")
    with open(file_path, 'r') as file:
        try:
            return json.load(file)
        except json.JSONDecodeError:
            raise ValueError("Invalid JSON format")

# Function to write a dictionary to a JSON file
def write_json(file_path, data):
    with open(file_path, 'w') as file:
        json.dump(data, file, indent=4)

# Function to calculate the square root of a number
def calculate_square_root(number):
    if number < 0:
        raise ValueError("Number must be non-negative")
    return math.sqrt(number)

# Function to check if a year is a leap year
def is_leap_year(year):
    if year % 4 == 0 and (year % 100 != 0 or year % 400 == 0):
        return True
    return False

# Function to count the number of vowels in a string
def count_vowels(s):
    vowels = "aeiouAEIOU"
    return sum(1 for char in s if char in vowels)

# Function to reverse a list
def reverse_list(lst):
    return lst[::-1]

# Function to calculate the factorial of a number iteratively
def factorial_iterative(n):
    if n < 0:
        raise ValueError("Number must be non-negative")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# Function to get the current timestamp
def get_current_timestamp():
    return datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")

# Function to merge two dictionaries
def merge_dicts(dict1, dict2):
    return {**dict1, **dict2}

# Main function to demonstrate the above functions
def main():
    try:
        length = 10
        width = 5
        print(f"Area of rectangle: {calculate_rectangle_area(length, width)}")
        print(f"Perimeter of rectangle: {calculate_rectangle_perimeter(length, width)}")

        number = 17
        print(f"Is {number} even? {is_even(number)}")

        string = "Hello Python"
        print(f"Uppercase: {to_uppercase(string)}")
        print(f"Number of vowels: {count_vowels(string)}")

        print(f"Square root of 16: {calculate_square_root(16)}")
        print(f"Factorial of 5: {factorial_iterative(5)}")

        year = 2024
        print(f"Is {year} a leap year? {is_leap_year(year)}")

        current_timestamp = get_current_timestamp()
        print(f"Current timestamp: {current_timestamp}")

        lst = [1, 2, 3, 4, 5]
        print(f"Reversed list: {reverse_list(lst)}")

        data = {"name": "John", "age": 30}
        file_path = "data.json"
        write_json(file_path, data)
        print(f"JSON written to {file_path}")

        read_data = read_json(file_path)
        print(f"Read JSON data: {read_data}")

    except Exception as e:
        print(f"An error occurred: {e}")

if __name__ == "__main__":
    main()
```
