# FIS (Fuzzy Inference System) #

#### 1. Import Library ####

```python
#Import Library
import numpy as np
import skfuzzy as fuzz
import matplotlib.pyplot as plt
```

#### 2. Deklarasi Variabel ####

```python
#Deklarasi variabel
neuroticism = np.arange (0, 40, 0.1)
st_neuroticism = np.arange (30, 65, 0.1)

extroversion = np.arange (10, 40, 0.1)
st_extroversion = np.arange (30, 65, 0.1)

SB_anxiety = np.arange (45, 50, 0.1)
SD_anxiety = np.arange (40, 60, 0.1)
Anxiety = np.arange (0, 60, 0.1)
```

#### 3. Deklarasi fungsi keanggotaan untuk setiap variable ####

```python
# range of membership function fuzzy set variable raw score of Neuroticism
neuroticism_A1 = fuzz.trimf(neuroticism, [0, 0, 10])
neuroticism_A2 = fuzz.trimf(neuroticism, [0, 10, 20])
neuroticism_A3 = fuzz.trimf(neuroticism, [10, 20, 30])
neuroticism_A4 = fuzz.trimf(neuroticism, [20, 30, 40])
neuroticism_A5 = fuzz.trimf(neuroticism, [30, 40, 40])

# range of membership function fuzzy set variable output standard score of neuroticism
neuroticism_B1 = fuzz.trimf(st_neuroticism, [30, 30, 41.67])
neuroticism_B2 = fuzz.trimf(st_neuroticism, [30, 41.67, 53.33])
neuroticism_B3 = fuzz.trimf(st_neuroticism, [41.67, 53.33, 65])
neuroticism_B4 = fuzz.trimf(st_neuroticism, [53.33, 65, 65])

# range of membership function fuzzy set variable raw score of Extroversion
extroversion_C1 = fuzz.trimf(extroversion, [10, 10, 17.5])
extroversion_C2 = fuzz.trimf(extroversion, [10, 17.5, 25])
extroversion_C3 = fuzz.trimf(extroversion, [17.5, 25, 32.5])
extroversion_C4 = fuzz.trimf(extroversion, [25, 32.5, 40])
extroversion_C5 = fuzz.trimf(extroversion, [32.5, 40, 40])

# range of membership function fuzzy set variable output standard score of Extroversion
extroversion_D1 = fuzz.trimf(st_extroversion, [30, 30, 41.67])
extroversion_D2 = fuzz.trimf(st_extroversion, [30, 41.67, 53.33])
extroversion_D3 = fuzz.trimf(st_extroversion, [41.67, 53.33, 65])
extroversion_D4 = fuzz.trimf(st_extroversion, [53.33, 65, 65])

# range of membership function fuzzy set variable raw score of SB_Anxiety (SB)
anxiety_SB1 = fuzz.trimf(SB_anxiety, [45, 45, 46.67])
anxiety_SB2 = fuzz.trimf(SB_anxiety, [45, 46.67, 48.33])
anxiety_SB3 = fuzz.trimf(SB_anxiety, [46.67, 48.33, 50])
anxiety_SB4 = fuzz.trimf(SB_anxiety, [48.33, 50, 50])

# range of membership function fuzzy set variable raw score of SD_Anxiety (SD)
anxiety_SD1 = fuzz.trimf(SD_anxiety, [40, 40, 46.67])
anxiety_SD2 = fuzz.trimf(SD_anxiety, [40, 46.67, 53.33])
anxiety_SD3 = fuzz.trimf(SD_anxiety, [46.67, 53.33, 60])
anxiety_SD4 = fuzz.trimf(SD_anxiety, [53.33, 60, 60])

# range of membership function fuzzy set variable output score of Anxiety
Anxiety_L = fuzz.trimf(Anxiety, [0, 0, 20])
Anxiety_M = fuzz.trimf(Anxiety, [0, 20, 40])
Anxiety_H = fuzz.trimf(Anxiety, [20, 40, 60])
Anxiety_E = fuzz.trimf(Anxiety, [40, 60, 60])
```

## A. Neuroticism ##

#### A.1. Menampilkan visualisasi fungsi keanggotaan Neuroticism ####

```python
# visualisasi membership function fuzzy set setiap variable
fig, (ax0, ax1) = plt.subplots(nrows=2, figsize=(15,10))
ax0.plot(neuroticism, neuroticism_A1, 'b', linewidth=1.5, label='A1')
ax0.plot(neuroticism, neuroticism_A2, 'g', linewidth=1.5, label='A2')
ax0.plot(neuroticism, neuroticism_A3, 'r', linewidth=1.5, label='A3')
ax0.plot(neuroticism, neuroticism_A4, 'y', linewidth=1.5, label='A4')
ax0.plot(neuroticism, neuroticism_A5, 'm', linewidth=1.5, label='A5')
ax0.set_title('Membership Function Fuzzy Set Variable Neuroticism')
ax0.legend()

ax1.plot(st_neuroticism, neuroticism_B1, 'b', linewidth=1.5, label='B1')
ax1.plot(st_neuroticism, neuroticism_B2, 'g', linewidth=1.5, label='B2')
ax1.plot(st_neuroticism, neuroticism_B3, 'r', linewidth=1.5, label='B3')
ax1.plot(st_neuroticism, neuroticism_B4, 'y', linewidth=1.5, label='B4')
ax1.set_title('Membership Function Fuzzy Set Variable Standard Score of Neuroticism')
ax1.legend()
```

#### A.2. Fuzzifikasi variable input Neuroticism ####

```python
# Fuzzifikasi variable input Neuroticism
neuro = 30
# neuro = 30

Neuroticism_A1_degree = fuzz.interp_membership(neuroticism, neuroticism_A1, neuro)
Neuroticism_A2_degree = fuzz.interp_membership(neuroticism, neuroticism_A2, neuro)
Neuroticism_A3_degree = fuzz.interp_membership(neuroticism, neuroticism_A3, neuro)
Neuroticism_A4_degree = fuzz.interp_membership(neuroticism, neuroticism_A4, neuro)
Neuroticism_A5_degree = fuzz.interp_membership(neuroticism, neuroticism_A5, neuro)

Neuroticism_fuzz_var = np.array([Neuroticism_A1_degree,
                                 Neuroticism_A2_degree,
                                 Neuroticism_A3_degree,
                                 Neuroticism_A4_degree,
                                 Neuroticism_A5_degree])
print('Fuzzy Variable ( Neuroticism =' , neuro, ') :', Neuroticism_fuzz_var)
```

#### A.3. Rule-based Neuroticism ####

```python
#Rule untuk Neoroticism
rule_neu1 = np.fmin(Neuroticism_A1_degree, neuroticism_B1)
rule_neu2 = np.fmin(Neuroticism_A1_degree, neuroticism_B2)
rule_neu3 = np.fmin(Neuroticism_A1_degree, neuroticism_B3)
rule_neu4 = np.fmin(Neuroticism_A1_degree, neuroticism_B4)

rule_neu5 = np.fmin(Neuroticism_A2_degree, neuroticism_B1)
rule_neu6 = np.fmin(Neuroticism_A2_degree, neuroticism_B2)
rule_neu7 = np.fmin(Neuroticism_A2_degree, neuroticism_B3)
rule_neu8 = np.fmin(Neuroticism_A2_degree, neuroticism_B4)

rule_neu9 = np.fmin(Neuroticism_A3_degree, neuroticism_B1)
rule_neu10 = np.fmin(Neuroticism_A3_degree, neuroticism_B2)
rule_neu11 = np.fmin(Neuroticism_A3_degree, neuroticism_B3)
rule_neu12 = np.fmin(Neuroticism_A3_degree, neuroticism_B4)

rule_neu13 = np.fmin(Neuroticism_A4_degree, neuroticism_B2)
rule_neu14 = np.fmin(Neuroticism_A4_degree, neuroticism_B3)
rule_neu15 = np.fmin(Neuroticism_A4_degree, neuroticism_B4)

rule_neu16 = np.fmin(Neuroticism_A5_degree, neuroticism_B3)
rule_neu17 = np.fmin(Neuroticism_A5_degree, neuroticism_B4)

# return array dengan bentuk dan ukuran yang sama dengan nilai elemen
# array diganti dengan nol
neuroticism_B = np.zeros_like(st_neuroticism)
```

```python
# visualisasi membership activity
fig, ax = plt.subplots(figsize=(15,5))

ax.fill_between(st_neuroticism, neuroticism_B, rule_neu1, facecolor='b', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B1, 'b', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu2, facecolor='r', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B2, 'r', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu3, facecolor='c', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B3, 'c', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu4, facecolor='m', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B4, 'm', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu5, facecolor='b', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B1, 'b', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu6, facecolor='r', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B2, 'r', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu7, facecolor='c', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B3, 'c', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu8, facecolor='m', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B4, 'm', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu9, facecolor='b', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B1, 'b', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu10, facecolor='r', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B2, 'r', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu11, facecolor='c', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B3, 'c', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu12, facecolor='m', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B4, 'm', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu13, facecolor='b', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B2, 'b', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu14, facecolor='c', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B3, 'c', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu15, facecolor='m', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B4, 'm', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu16, facecolor='c', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B3, 'c', linewidth=1.5, linestyle='--')
ax.fill_between(st_neuroticism, neuroticism_B, rule_neu17, facecolor='m', alpha=0.7)
ax.plot(st_neuroticism, neuroticism_B4, 'm', linewidth=1.5, linestyle='--')

ax.set_title('Visualisasi Membership Activity Neuroticism')

for ax in (ax,):
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.get_xaxis().tick_bottom()
    ax.get_yaxis().tick_left()
    plt.tight_layout()
```

#### A.5. Agregasi Neuroticism ####

```python
# Tahap Aggregation
aggregated_neuro = np.fmax(rule_neu1,
                     np.fmax(rule_neu2,
                        np.fmax(rule_neu3,
                            np.fmax(rule_neu4,
                                np.fmax(rule_neu5,
                                    np.fmax(rule_neu6,
                                        np.fmax(rule_neu7,
                                            np.fmax(rule_neu8,
                                                np.fmax(rule_neu9,
                                                    np.fmax(rule_neu10,
                                                        np.fmax(rule_neu11,
                                                            np.fmax(rule_neu12,
                                                                np.fmax(rule_neu13,
                                                                    np.fmax(rule_neu14,
                                                                        np.fmax(rule_neu15,
                                                                            np.fmax(rule_neu16, rule_neu17))))))))))))))))
```
